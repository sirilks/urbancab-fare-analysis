```python
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    
    print(f"{col}: {len(outliers)} outliers")
```

### Demand by Hour

```python
# hourly values
df['Hour_of_Day'].value_counts().sort_index()
```

```python
# total demand for each hour
hourly_demand = (
    df.groupby('Hour_of_Day')['Ride_Requests']
      .sum()
      .reset_index()
)

hourly_demand
```

```python
# peak demand hour
peak_hour = hourly_demand.loc[
    hourly_demand['Ride_Requests'].idxmax()
]

print("Peak Hour:", peak_hour['Hour_of_Day'])
print("Ride Requests:", peak_hour['Ride_Requests'])
```

```python
# top 5 peak hours
top_hours = (
    hourly_demand
    .sort_values('Ride_Requests', ascending=False)
    .head(5)
)

top_hours
```

```python
def time_period(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['Time_Period'] = df['Hour_of_Day'].apply(time_period)
```

```python
df[['Hour_of_Day', 'Time_Period']].head()
```

```python
# hours with high demand and low driver availability
hourly_supply = (
    df.groupby('Hour_of_Day')
      .agg(
          ride_requests=('Ride_Requests', 'sum'),
          avg_active_drivers=('No_of_active_drivers', 'mean'),
          avg_demand_supply_ratio=('Demand_Supply_Ratio', 'mean')
      )
      .reset_index()
)

hourly_supply
```

```python
# hours with high demand AND low driver availability
hourly_supply.sort_values(
    ['ride_requests', 'avg_active_drivers'],
    ascending=[False, True]
).head(10)
```

```python
hourly_supply.sort_values(
    'avg_demand_supply_ratio',
    ascending=False
).head(10)
```

### Weekday vs Weekend

```python
df['Day_of_Week'].value_counts()
```

```python
# demand by day
daily_demand = (
    df.groupby('Day_of_Week')['Ride_Requests']
      .sum()
      .reset_index()
)

daily_demand
```

```python
# days in the correct order
day_order= [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

daily_demand['Day_of_Week']=pd.Categorical(
    daily_demand['Day_of_Week'],
    categories=day_order,
    ordered=True
)

daily_demand=daily_demand.sort_values('Day_of_Week')
daily_demand
```

```python
# Visualize demand
plt.figure(figsize=(12, 6))

sns.barplot(
    data=daily_demand,
    x='Day_of_Week',
    y='Ride_Requests'
)

plt.title('Ride Demand by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Total Ride Requests')
plt.xticks(rotation=45)
plt.show()
```

```python
# Compare weekday vs weekend
df['Day_Type'] = df['Day_of_Week'].apply(
    lambda x: 'Weekend' if x in ['Saturday', 'Sunday'] else 'Weekday'
)

weekday_weekend = (
    df.groupby('Day_Type')
      .agg(
          total_ride_requests=('Ride_Requests', 'sum'),
          average_ride_requests=('Ride_Requests', 'mean'),
          average_active_drivers=('No_of_active_drivers', 'mean'),
          average_demand_supply_ratio=('Demand_Supply_Ratio', 'mean')
      )
      .reset_index()
)

weekday_weekend
```

```python
# Visualize Weekday vs Weekend

plt.figure(figsize=(8, 5))

sns.barplot(
    data=weekday_weekend,
    x='Day_Type',
    y='total_ride_requests'
)

plt.title('Weekday vs Weekend Ride Demand')
plt.xlabel('Day Type')
plt.ylabel('Total Ride Requests')
plt.show()
```

```python
# Highest-demand day

highest_demand_day = daily_demand.loc[
    daily_demand['Ride_Requests'].idxmax()
]

print("Highest Demand Day:", highest_demand_day['Day_of_Week'])
print("Ride Requests:", highest_demand_day['Ride_Requests'])
```
